```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get uploaded file name
filename = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

# Input gamma value
gamma = float(input("Enter the value of gamma: "))

# Normalize image to range [0, 1]
img_normalized = img / 255.0

# Power Law / Gamma Transformation
# s = c * r^gamma, where c = 1
enhanced = np.power(img_normalized, gamma)

# Convert back to range [0, 255]
enhanced = np.uint8(enhanced * 255)

# Display Original and Enhanced Image
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(enhanced, cmap='gray')
plt.title("Gamma Transformed Image")
plt.axis("off")

plt.show()
```
